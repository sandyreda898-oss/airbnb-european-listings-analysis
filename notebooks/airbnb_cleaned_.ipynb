```python
df = df.rename(columns={
    "City": "city",
    "Property Name": "property_name",
    "Price": "price_raw",
    "Rating": "rating",
    "Number of Reviews": "number_of_reviews",
    "Superhost Status": "superhost",
    "Guest Favorite Status": "guest_favorite",
    "Property URL": "property_url"
})
df
```

```python
for col in ["city", "property_name", "property_url"]:
    df[col] = (
        df[col]
        .astype("string")
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )

df["city"] = df["city"].str.title()
```

```python
df[["city", "property_name", "property_url"]]
```

```python
def to_bool(x):
    if pd.isna(x):
        return pd.NA
    if isinstance(x, bool):
        return x
    s = str(x).strip().lower()
    if s in ["true", "1", "yes"]:
        return True
    if s in ["false", "0", "no"]:
        return False
    return pd.NA

for col in ["superhost", "guest_favorite"]:
    df[col] = df[col].map(to_bool).astype("boolean")
```

```python

def parse_price(x):
    if pd.isna(x):
        return pd.Series([np.nan, np.nan])

    s = str(x).strip()
    if s == "" or s.lower() == "nan":
        return pd.Series([np.nan, np.nan])

    currency = np.nan
    if "$" in s:
        currency = "USD"
    elif "€" in s or "EUR" in s.upper():
        currency = "EUR"
    elif "£" in s:
        currency = "GBP"

    m = re.search(r"(\d[\d,]*(?:\.\d+)?)", s)
    amount = float(m.group(1).replace(",", "")) if m else np.nan

    return pd.Series([amount, currency])

df[["price_amount", "price_currency"]] = df["price_raw"].apply(parse_price)

df["price_suspicious"] = (
    df["price_amount"].notna() &
    ((df["price_amount"] < 10) | (df["price_amount"] > 10000))
)
```

```python
df["property_name_key"] = (
    df["property_name"]
    .str.lower()
    .str.replace(r"[^\w\s]", "", regex=True)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)
```

```python
df["rating"] = pd.to_numeric(df["rating"], errors="coerce")
df.loc[(df["rating"] < 0) | (df["rating"] > 5), "rating"] = np.nan
```

```python
df["number_of_reviews"] = pd.to_numeric(
    df["number_of_reviews"], errors="coerce"
).astype("Int64")

df.loc[df["number_of_reviews"] < 0, "number_of_reviews"] = pd.NA

df["reviews_outlier_flag"] = df["number_of_reviews"] > 5000
```

```python
df["listing_id"] = df["property_url"].str.extract(r"/rooms/(\d+)")[0]
```

```python
print("Exact duplicates:", df.duplicated().sum())
print("Duplicate URLs:", df.duplicated(subset=["property_url"]).sum())

df = df.drop_duplicates()
df = df.drop_duplicates(subset=["property_url"], keep="first")
```

```python
dup_names = df[df.duplicated(subset=["city", "property_name_key"], keep=False)]
print(dup_names[["city", "property_name", "property_url"]])
```

```python
missing_report = df.isna().sum().sort_values(ascending=False)
print(missing_report)
```

```python
df = df.dropna(subset=["city", "property_url"])
```

```python
mask = df["price_amount"].isna() & df["price_raw"].notna()
print(df.loc[mask, ["city", "property_name", "price_raw", "price_currency"]])
```

```python
df["price_parse_failed"] = df["price_raw"].notna() & df["price_amount"].isna()

print(df["price_parse_failed"].sum(), "rows with unparseable price")
```

```python
print(df.loc[df["price_suspicious"], ["city","property_name","price_raw","price_amount","price_currency"]].to_string())

print(df.loc[df["reviews_outlier_flag"], ["city","property_name","number_of_reviews","rating","property_url"]].to_string())

print("dupURL:", df.duplicated(subset=["property_url"]).sum(),
      "| dupID:", df.duplicated(subset=["listing_id"]).sum(),
      "| dupCityName:", df.duplicated(subset=["city","property_name_key"]).sum(),
      "| dupFullRow:", df.duplicated().sum())
```

```python
df["price_is_per_night"] = (
    df["price_raw"]
    .astype("string")
    .str.contains(r"\bnight\b", case=False, na=False)
    .fillna(False)
    .astype("boolean")
)

df["price_suspicious"] = (
    df["price_suspicious"].fillna(False)
    | df["price_is_per_night"].fillna(False)
).astype("boolean")

print("Suspicious now:", int(df["price_suspicious"].sum()))
```
